```python
from pathlib import Path
import pandas as pd
DATASET_PATH = Path("../data/dataset_profesor.csv")
df = pd.read_csv(DATASET_PATH)
```

- __Variables__/características: `feature_columns`

- __Data leakage__: `leakage_column`

- Variable __objetivo__: `target_column`

Con `leakage_column` señalamos de forma explícita cuál es data leakage, pero podríamos simplemente no incluirla en nuestra X.

```python
feature_columns = [
    "tenure_months",
    "monthly_spend",
    "support_tickets_3m",
    "usage_change_last_month",
    "late_payments_6m",
]
target_column = "churn_next_month"
leakage_column = "cancellation_request_registered"
```

#### Simplemente separamos columnas

```python
X = df[feature_columns]
y = df[target_column]
```

```python
X.head()
```

```python
y.head()
```

## `.fit`

Tomamos un modelo: `DecisionTreeClassifier`. De momento no estudiamos cómo trabaja por dentro. Seleccionamos una semilla.

```python
from sklearn.tree import DecisionTreeClassifier
RANDOM_STATE = 47
model = DecisionTreeClassifier(random_state=RANDOM_STATE)
model.fit(X,y)
```

```python
model.predict(X)
```

Como no hemos separado entre train/val/test, claramente predice lo mismo.

```python
y.values
```

## `accuracy_score`

Veamos cuál es el accuracy de nuestro modelo.

```python
from sklearn.metrics import accuracy_score
accuracy_score(
    y_true=y, 
    y_pred=model.predict(X)
    )
```

Claramente predice bien el 100% de los casos, porque estamos testeando unos datos sobre los que ha sido entrenado. Data leakage de manual.

## `train_test_split`

Separemos los datos en train/validation/test. `train_test_split` lo separa en dos conjuntos (que no necesariamente es train y test).

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    train_size=0.8,
    random_state=RANDOM_STATE,
    shuffle=True,
    stratify=y
)
```
